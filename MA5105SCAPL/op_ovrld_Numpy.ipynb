### Let OOP solve the problem

We can create a `Vector` class that knows how to perform

$$
\alpha\mathbf{x}
\qquad\text{and}\qquad
\mathbf{x}+\mathbf{y}.
$$

```python
class TBTVEC:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2
```

```python
x = TBTVEC(1, 2)
y = TBTVEC(10, 20)
```

- What should $$ x + y $$ mean?

```python
# z = x + y      # there to show the failure, uncomment to see
```

## Can we teach Python new operations ? 

- offcourse we can, duh !!!

- Using methods 

```python
class TBTVEC2:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2

    def add(self, other):
        return TBTVEC2(
        self.x1 + other.x1,
        self.x2 + other.x2)

    def scalar_multiply(self, alpha):
        return TBTVEC2(
        alpha * self.x1,
        alpha * self.x2)
```

```python
# Now I am ready to add the two vectors x and y and mulitply by scalar
x = TBTVEC2(1, 2)
y = TBTVEC2(10, 20)

# x + y 
z = x.add(y)

print (z.x1)
print (z.x2)

# 2 * X
L = x.scalar_multiply(2)
print(L.x1)
print(L.x2)

# 2 * x + y

result = L.add(y)
print(result.x1)
print(result.x2)
```

# Operator Overloading

- Python allows a class to define how operators behave when they are applied to its objects.

- Operator overloading allows us to define how standard Python operators behave for objects of a class.

- Operators such as +, -, *, /, and == can be given class-specific meanings.

- Python implements these operators through special methods, also called dunder methods.

- For example:
    - x + y → x.__add__(y)
    - x * y → x.__mul__(y)
    - x == y → x.__eq__(y)

- This allows objects to behave naturally according to the mathematical meaning we want.

- Why These Names

    - `__add__(y)`
    - `__mul__(y)`
    - `__eq__(y)`

Matter ??

- Python's built-in objects already have these operators implemented using the same special method names.

```python
3 + 4
```

- is essentially handled through:

```python
(3).__add__(4)
```

```python
int.__add__(3, 4)
```

- Python therefore gives special meaning to method names such as:
    - `__add__` (+ operator)
    - `__sub__` (- operator)
    - `__mul__` (* operator)
    - `__truediv__`          (/ operator)
    - `__eq__` (= operator)
    - `__lt__` (< operator)

- These names are not arbitrary. 
- Python's interpreter recognizes them and uses them when the corresponding operators appear.
- Think of these special method names as part of Python's protocol.
- So when we define

```python
# def __add__(self, other):
```

- we aren't simply creating a method called `__add__`.

- We are telling Python:
    - **When someone writes x + y and x is one of my objects, this is what + should do.**

## Lets do it with 2 X 1 Vec class 

```python
class TBTVEC3:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2

    def __add__(self, other):
        return TBTVEC3(
            self.x1 + other.x1,
            self.x2 + other.x2
        )

    def __mul__(self, alpha):
        return TBTVEC3(
            alpha * self.x1,
            alpha * self.x2
        )

    def __rmul__(self, alpha):
        return self.__mul__(alpha)
```

```python
x = TBTVEC3(1, 2)
y = TBTVEC3(10, 20)
z = x * 2 + y
```

```python
print(z.x1)
print(z.x2)
```

```python
z = 2 * x + y
```

## Linear algebra and other advanced math

- Internally, NumPy doesn’t invent its own math routines: it relies on BLAS and LAPACK to do this kind of math - the same as many other languages.

- NumPy provides linear algebra operations through the `np.linalg` submodule

- Solving a system of linear equations
- Consider:
  
    $ 2 x + y = 5$
  
    $ x + 3 y = 6$.

```python
# Matrix and vectors
A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 6])
```

```python
# Solve A x = b
x = np.linalg.solve(A, b)
```

```python
print("A =")
print(A)
```

```python
print("\nb =")
print(b)

print("\nSolution x =")
print(x)
```

```python
# Check the solution
print("\nCheck A @ x =")
print(A @ x)
```

- Matrix transpose

```python
# Transpose
AT = A.T
AT
```

- Determinant

```python
detA = np.linalg.det(A)
print("detA = ", detA)
```

- Inverse

```python
Ainv = np.linalg.inv(A)
Ainv
```

- Norm of a vector

```python
norm_x = np.linalg.norm(x)
norm_x
```

- Eigenvalues and eigenvectors

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
```

```python
print("||x|| =", norm_x)
print("Eigenvalues =", eigenvalues)
print("Eigenvectors =\n", eigenvectors)
```
