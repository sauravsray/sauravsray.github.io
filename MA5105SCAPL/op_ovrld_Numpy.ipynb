# From Python Lists to Numerical Arrays

- Suppose we want to perform a simple vector operation:

$$ \mathbf{x} = (1,2)^T
\qquad
\mathbf{y} = (10, 20)^T$$



- Using Python lists:

```python
x = [1, 2]
y = [10, 20]
```

- We might want

$$ \mathbf{z}=2\mathbf{x}+\mathbf{y} $$

- But Python does not treat lists as mathematical vectors:

```python
2*x + y
```

- What did it do ??

- does $\textbf{not}$ perform vector arithmetic.

- How do we achieve what we want ?

- `*` does perform vector arithmetic. Same goes true for `+` as well 

- **Python does not know that `x` and `y` are mathematical vectors. It only knows that they are `list` objects.**

That's the conceptual point.

### Let OOP solve the problem

We can create a `Vector` class that knows how to perform

$$
\alpha\mathbf{x}
\qquad\text{and}\qquad
\mathbf{x}+\mathbf{y}.
$$

```python
class TBTVEC:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2
```

```python
x = TBTVEC(1, 2)
y = TBTVEC(10, 20)
```

- What should $$ x + y $$ mean?

```python
# z = x + y      # there to show the failure, uncomment to see
```

## Can we teach Python new operations ? 

- offcourse we can, duh !!!

- Using methods 

```python
class TBTVEC2:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2

    def add(self, other):
        return TBTVEC2(
        self.x1 + other.x1,
        self.x2 + other.x2)

    def scalar_multiply(self, alpha):
        return TBTVEC2(
        alpha * self.x1,
        alpha * self.x2)
```

```python
# Now I am ready to add the two vectors x and y and mulitply by scalar
x = TBTVEC2(1, 2)
y = TBTVEC2(10, 20)

# x + y 
z = x.add(y)

print (z.x1)
print (z.x2)

# 2 * X
L = x.scalar_multiply(2)
print(L.x1)
print(L.x2)

# 2 * x + y

result = L.add(y)
print(result.x1)
print(result.x2)
```

### This works, but it does not look like the mathematics.

# Operator Overloading

- Python allows a class to define how operators behave when they are applied to its objects.

- Operator overloading allows us to define how standard Python operators behave for objects of a class.

- Operators such as +, -, *, /, and == can be given class-specific meanings.

- Python implements these operators through special methods, also called dunder methods.

- For example:
    - x + y → x.__add__(y)
    - x * y → x.__mul__(y)
    - x == y → x.__eq__(y)

- This allows objects to behave naturally according to the mathematical meaning we want.

- Why These Names

    - `__add__(y)`
    - `__mul__(y)`
    - `__eq__(y)`

Matter ??

- Python's built-in objects already have these operators implemented using the same special method names.

```python
3 + 4
```

- is essentially handled through:

```python
(3).__add__(4)
```

```python
int.__add__(3, 4)
```

- Python therefore gives special meaning to method names such as:
    - `__add__` (+ operator)
    - `__sub__` (- operator)
    - `__mul__` (* operator)
    - `__truediv__`          (/ operator)
    - `__eq__` (= operator)
    - `__lt__` (< operator)

- These names are not arbitrary. 
- Python's interpreter recognizes them and uses them when the corresponding operators appear.
- Think of these special method names as part of Python's protocol.
- So when we define

```python
# def __add__(self, other):
```

- we aren't simply creating a method called `__add__`.

- We are telling Python:
    - **When someone writes x + y and x is one of my objects, this is what + should do.**

## Lets do it with 2 X 1 Vec class 

```python
class TBTVEC3:

    def __init__(self, x1, x2):
        self.x1 = x1
        self.x2 = x2

    def __add__(self, other):
        return TBTVEC3(
            self.x1 + other.x1,
            self.x2 + other.x2
        )

    def __mul__(self, alpha):
        return TBTVEC3(
            alpha * self.x1,
            alpha * self.x2
        )

    #def __rmul__(self, alpha):
    #    return self.__mul__(alpha)
```

```python
x = TBTVEC3(1, 2)
y = TBTVEC3(10, 20)
z = x * 2 + y
```

```python
print(z.x1)
print(z.x2)
```
